## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


A fixed rule could prioritize every page whose impressions declined by more than a chosen threshold. However, a single threshold may overlook differences in content age, recent clicks, update history, and traffic volume.

A ranking approach could combine several signals to help content strategists identify pages worth reviewing first. Its usefulness must be compared against a simple rule-based baseline and validated using human-reviewed priorities. If the ranking does not improve the decision, a simpler rule may be sufficient.

The current dataset supports an initial investigation, but it does not by itself prove that ML will outperform a fixed rule.
