```python
import pandas as pd

csv_path = "/content/Flyrank-intern-repo/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(csv_path)

print("Total pages:", len(df))
print(
    "Declining-page percentage:",
    round((df["trend_direction"] == "down").mean() * 100, 2),
    "%"
)
print(
    "Search volume vs impressions correlation:",
    round(df["search_volume"].corr(df["impressions_90d"]), 3)
)
```
